```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("Personal_Data_Sheets.csv")
print(data.shape)
data.head()
```

# Maps Types

## All

```python
output = data.groupby("Map_Type").agg({"Won":['mean', 'count']})
output.columns = ["Win Rate", "Games"]
output["Win Rate"] = (output["Win Rate"] * 100)#.info()
output.sort_values("Win Rate", ascending=False)
```

## Unranked

```python
output = data[~data["Ranked"]].groupby("Map_Type").agg({"Won":['mean', 'count']})
output.columns = ["Win Rate", "Games"]
output["Win Rate"] = (output["Win Rate"] * 100).astype(float).round(2)
output.sort_values("Win Rate", ascending=False)
```

## Ranked

```python
output = data[data["Ranked"]].groupby("Map_Type").agg({"Won":['mean', 'count']})
output.columns = ["Win Rate", "Games"]
output["Win Rate"] = (output["Win Rate"] * 100).astype(float).round(2)
output.sort_values("Win Rate", ascending=False)
```
